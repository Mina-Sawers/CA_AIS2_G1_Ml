https://leetcode.com/problems/basic-calculator-ii/description/?envType=problem-list-v2&envId=math

```python
s = " 3/2 "
nums = []
current_num = 0
last_op = '+'
operations = ['+','-','*','/']
slen = len(s)-1
for i,c in enumerate(s):
    if c not in operations:
        if c.isdigit():
            current_num = current_num*10 + int(c)

        if last_op == '+' and (i == slen or s[i+1] in operations):
            nums.append(current_num)
        elif last_op == '-' and (i == slen or s[i+1] in operations):
            nums.append(-current_num)
        elif last_op == '*' and (i == slen or s[i+1] in operations):
            nums.append(nums.pop() * current_num )
        elif last_op == '/' and (i == slen or s[i+1] in operations):
            nums.append(int(nums.pop() / current_num))

    elif c in operations:
        if c=='+':
            last_op = '+'
        elif c=='-':
            last_op = '-'
        elif c=='*':
            last_op = '*'
        elif c=='/':
            last_op = '/'

        current_num = 0



print(sum(nums))
```

https://leetcode.com/problems/search-insert-position/description/

```python
nums=[1,3,5,6]
target = 100
def searchInsert(nums,target):
    left = 0
    right = len(nums)-1
    mid = 0
    while left<=right:
        mid = (left+right) // 2

        if nums[mid]==target:
            return mid
        elif target>nums[mid]:
            left = mid+1
        elif target<nums[mid]:
            right = mid-1

    return left
```
