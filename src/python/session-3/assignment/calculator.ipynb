```python
print("Welcome to the simple Calculator")
print("Select an operation\n1. Addition(+)\n2. Subtraction (-)\n3. Multiplication\n4. Division(/)")

while True:
    choice = input("Enter your choice (1/2/3/4) or 'exit' to quit: ").lower()
    if(choice=='exit'):
        print("Thank You!")
        break
    try:
        num1 = float(input("Enter the 1st number: "))
        num2 = float(input("Enter the 2nd number: "))
    except ValueError:
        print("Invalid Input, Enter numeric values only.")
        continue
    result = 0
    match choice:
        case '1':
            result = num1+num2
        case '2':
            result = num1-num2
        case '3':
            result = num1*num2
        case '4':
            if num2==0:
                print("Cannot divide by zero")
                continue
            result = num1/num2
        case _:
            print('Invalid Choice, Kindly choose 1,2,3,4')
            continue
    print(f"Result = {result}")
    user_input = input("Enter y or Y to continue and n or N to break")
    if(user_input=='n' or user_input=='N'):
        break
```
