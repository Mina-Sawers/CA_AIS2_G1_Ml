```python
def machine_learning(model_name,*args,**hyperparameters):
    '''
    This is a test function
    Args:
        model_name: this is the model name
        args: these are the models
        hyperparameters: these are the hyperparameters
    Returns:
        It returns nothing
    '''
    print(model_name)
    for arg in args:
        print(arg)
    for key,value in hyperparameters.items():
        print(key+" "+value)
    
```
