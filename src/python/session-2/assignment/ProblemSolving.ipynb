```python
def are_you_playing_banjo(name):
    if(name[0]=='R' or name[0]=='r'):
        return name+" plays banjo"
    else:
        return name+" does not play banjo"
print(are_you_playing_banjo("ramez"))
```

```python
def reverse_words(s):
    list1 = s.split()
    list1 = list1[::-1]
    new_s = " ".join(list1)
    return new_s
reverse_words("hello world!")
```
