```python
class Student:
    def __init__(self,name,age)->None:
        self._name = name
        self._age = age

    def get_info(self):
        return f"my name is {self.name}, and ge is {self.age}"

    @property
    def name(self):
        return self._name
    @property
    def age(self):
        return self._age

    @age.setter
    def age(self,value):
        if value<0:
            raise ValueError("age must be positive")
        self._age = value
    
s1 = Student("mina",14)
print(s1.name)
s1.age = 10
print(s1.age)
```

```python
class Shape:
    def calculate_area(self):
        pass

class Circle(Shape):
    def __init__(self,radius):
        self.radius = radius
        # super().__init__()
    def calculate_area(self):
        return 3.14*(self.radius**2)
        # return super().calculate_area()
class Square(Shape):
    def __init__(self,side):
        # super().__init__()
        self.side = side
    def calculate_area(self):
        return self.side**2

if __name__ == "__main__":
    x = Circle(5)
    y = Square(4)
    print("Area: ",x.calculate_area())
    print("Area: ",y.calculate_area())

    
```
