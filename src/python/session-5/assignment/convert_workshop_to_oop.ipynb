```python
class NumbersManipulation:

    def __init__(self,num):
        self.num = num

    def factorial(self)->int:
        num = self.num
        ''' 
        calculate n! using Recursion function
        Args:
            num(int): user input the int number
        Return:
            num(int): return the factorial
        Examples:
            factorial(5) = 120
        '''
        if num==0:
            return 1
        return num*factorial(num-1)

    def is_prime(self)->bool:
        num = self.num
        ''' 
        This function check whether the number prime or not

        Args:
            num(int): number to check
        Return:
            bool: True if the number is prime or false ifnot prime
        Examples:
            is_prime(7)->True
        '''
        if num<2:
            return False
        for i in range(2,int(num**0.5)+1):
            if num%i==0:
                return False
        return True

n = NumbersManipulation(7)
print(n.factorial())
print(n.is_prime())
```

```python
class CommonDivisor:
    def __init__(self,num1,num2):
        self.num1 = num1
        self.num2 = num2
        
    def common_divisor(self)->list[int]:
        num1 = self.num1
        num2 = self.num2
        ''' 
        this function help to calc the common divisors between two numbers

        
        '''
        limit = min(num1,num2)
        divs = []
        for divisor in range(1,limit+1):
            if num1%divisor==0 and num2%divisor==0:
                divs.append(divisor)
        return divs 
cd = CommonDivisor(20,40)
print(cd.common_divisor())
```

```python
class ReverseTxt:
    def __init__(self,s):
        self.s = s

    def reverse_txt(self)->list[str]:
        s = self.s
        if not isinstance(s,str):
            raise TypeError("text must be a string")
        lst = s.split()
        #print(lst)
        new_list = lst[::-1]
        #print(new_list)
        return new_list
    def reverse_word(self)->str:
        s = self.s
        return " ".join(s.split()[::-1])
    
rt = ReverseTxt("Hello my name is Mina")
print(rt.reverse_txt())
print(rt.reverse_word())
```

```python
class Dict:
    def __init__(self,lst):
        self.lst = lst
    def generate_dict(self)->dict:
        lst = self.lst
        ''' 
        This function to return count of unique words
        '''
        dict1 = {}
        for item in lst:
            item = item.lower()
            if item in dict1:
                dict1[item]+=1
            else:
                dict1[item]=1
        return dict1
    
d = Dict(["Welcome", "Ali", "Hi", "Ali", "No", "Hi", "No", "Ali", "No", "Ali"])
print(d.generate_dict())
```
