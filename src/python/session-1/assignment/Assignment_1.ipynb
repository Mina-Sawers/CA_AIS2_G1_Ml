# Task 2

```python
def keep_letters_and_spaces(text):
    clean_text = [char for char in text if char.isalpha() or char.isspace()]
    return "".join(clean_text)
def first_word_reversed(text):
    return text.split(" ")[0][::-1]
def get_second_word(text):
    return text.split(" ")[1]
```

```python
encoded_message = "###!!@mocleW EPGTQ!!!6789"
clean_message = keep_letters_and_spaces(encoded_message)
print(clean_message)
```

```python
first_reversed = first_word_reversed(clean_message)+"e"
print(first_reversed)
```

```python
second_word = get_second_word(clean_message)
second_word = second_word.replace("E","")
print(second_word)
```

```python
print(first_reversed+" "+second_word)
```
