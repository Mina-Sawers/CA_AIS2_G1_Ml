```python
def common_divisor(num1:int,num2:int)->list[int]:
    ''' 
    this function help to calc the common divisors between two numbers

    
    '''
    limit = min(num1,num2)
    divs = []
    for divisor in range(1,limit+1):
        if num1%divisor==0 and num2%divisor==0:
            divs.append(divisor)
    return divs 
```

```python
print(common_divisor(10,20))
```

```python
def reverse_txt(s:str)->list[str]:
    if not isinstance(s,str):
        raise TypeError("text must be a string")
    lst = s.split()
    #print(lst)
    new_list = lst[::-1]
    #print(new_list)
    return new_list
```

```python
s = "hello depi"
print(reverse_txt(s))
```

```python
def reverse_word(s:str)->str:
    return " ".join(s.split()[::-1])
print(reverse_word("hi depi"))
```

```python
import random
def generate_password(length:int)->str:
    ''' 
    This function returns random password of certain length
    
    '''
    characters = "abcdefghijklmnopqrstuvwxyzABCDEFGHIJKLMNOPQRSTUVWXYZ0123456789!"#$%&\'()*+,-./:;<=>"
    password = ""
    for _ in range(length):
        password+=random.choice(characters)
    return password
```

```python
def generate_dict(lst:list)->dict:
    ''' 
    This function to return count of unique words
    '''
    dict1 = {}
    for item in lst:
        item = item.lower()
        if item in dict1:
            dict1[item]+=1
        else:
            dict1[item]=1
    return dict1
generate_dict(["Welcome", "Ali", "Hi", "Ali", "No", "Hi", "No", "Ali", "No", "Ali"])
```
