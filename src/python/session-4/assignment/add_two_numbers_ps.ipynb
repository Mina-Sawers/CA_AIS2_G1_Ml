```python
from typing import Optional
class ListNode:
    def __init__(self, val=0, next=None):
        self.val = val
        self.next = next
class Solution:
    def addTwoNumbers(self, l1: Optional[ListNode], l2: Optional[ListNode]) -> Optional[ListNode]:
        current1 = l1
        current2 = l2
        s1 = ""
        s2 = ""
        while current1 is not None:
            s1+=str(current1.val)
            current1 = current1.next
        while current2 is not None:
            s2+=str(current2.val)
            current2 = current2.next
        num1 = int(s1[::-1])
        num2 = int(s2[::-1])
        result = num1+num2
        result = str(result)
        
        dummy = ListNode(0)
        current = dummy
    
        result = result[::-1]


        for c in result:
            new_node = ListNode(int(c))
            current.next = new_node
            current = current.next
            
            

        return dummy.next
    
# 1. Manually build the first linked list: 2 -> 4 -> 3
l1_node1 = ListNode(2)
l1_node2 = ListNode(4)
l1_node3 = ListNode(3)
l1_node1.next = l1_node2 # Link 2 to 4
l1_node2.next = l1_node3 # Link 4 to 3

# 2. Manually build the second linked list: 5 -> 6 -> 4
l2_node1 = ListNode(5)
l2_node2 = ListNode(6)
l2_node3 = ListNode(4)
l2_node1.next = l2_node2 # Link 5 to 6
l2_node2.next = l2_node3 # Link 6 to 4

my_solution = Solution()
result = my_solution.addTwoNumbers(l1_node1,l2_node1)

while result is not None:
    print(result.val)
    result = result.next


        
```
