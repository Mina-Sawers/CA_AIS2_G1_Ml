```python
def setZeroes(matrix: list[list[int]]) -> None:
    n_rows = len(matrix)
    n_cols = len(matrix[0])
    zero_rows = set()
    zero_cols = set()

    for i in range(0,n_rows):
        for j in range(0,n_cols):
            if matrix[i][j]==0:
                zero_rows.add(i)
                zero_cols.add(j)

    for row in zero_rows:
        for j in range(0,n_cols):
            matrix[row][j] = 0

    for col in zero_cols:
        for i in range(0,n_rows):
            matrix[i][col]=0
    

matrix =[[1,1,1],[1,0,1],[1,1,1]]
setZeroes(matrix)
print(matrix)
```
