```python
import numpy as np
import matplotlib.pyplot as plt

x = np.array([1,2,3,4])
y = np.array([2,2.8,3.6,4.5])

n = len(x)
```

```python
w = 0
b = 0

learning_rate = 0.01
epochs = 1000
cost_value = []

print(f"Initial w:{w:.4f}")
print(f"Initial b:{b:.4f}")
print(f"Initial learning rate:{learning_rate:.4f}")

for i in range(epochs):
    y_hat = w*x+b
    error = y_hat-y
    #=============
    D_w = np.mean(error*x)
    D_b = np.mean(error)

    w = w - learning_rate*D_w
    b = b - learning_rate*D_b

    y_hat = w*x+b
    cost = 0.5*np.mean((y_hat-y)**2)
    cost_value.append(cost)

    if (i+1) %100 == 0 or i == 0:
        print(f"Epoch {i +1}: "
              f"cost = {cost}, "
              f"w = {w}"
              f"b ={b}"
              
              
              )
```

```python
plt.figure(figsize=(12,5))
plt.subplot(1,2,1)
plt.scatter(x,y)
plt.plot(x,w*x+b,color='r')
```

```python
plt.subplot(1,2,1)
plt.plot(cost_value)
```
